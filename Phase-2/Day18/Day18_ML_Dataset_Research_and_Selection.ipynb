## 3️⃣ Final Dataset Selection

**Selected: Crop Recommendation Dataset**

**Why this one:**
- It is the internship's assigned long-term project (soil/weather-based crop and fertilizer recommendation), so building on it now carries directly into every later phase instead of starting a separate track.
- It is already proven end-to-end: Decision Tree beat Logistic Regression and KNN on it (Day 12), it was evaluated with a confusion matrix and weighted precision/recall/F1 (Day 14), and hyperparameters were tuned with GridSearchCV/RandomizedSearchCV (Day 15).
- Its 7-column schema (N, P, K, temperature, humidity, ph, rainfall) maps cleanly onto a simple frontend form (Phase 03), a simple database table (Phase 04), and a simple model input at inference time (Phase 05) — no messy categorical fields to encode, unlike the Fertilizer dataset.
- Compared to the Yield dataset (regression, only 2 usable numeric features), it gives a richer, more demonstrable classification problem for a portfolio project.

## Basic Operations on the Selected Dataset

Quick exploratory checks on the Crop Recommendation Dataset before it becomes the Days 19–20 individual project model.

> Place `Crop_recommendation.csv` in the same folder as this notebook (or update `DATA_PATH` below) before running the cells.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = "Crop_recommendation.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

### Shape and structure

```python
print("Shape:", df.shape)
df.info()
```

### Summary statistics

```python
df.describe()
```

### Missing value check

```python
df.isnull().sum()
```

### Class distribution (22 crop labels)

```python
print("Number of classes:", df['label'].nunique())

plt.figure(figsize=(10, 5))
df['label'].value_counts().plot(kind='bar')
plt.title("Number of samples per crop")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

### Feature correlation heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.drop(columns=['label']).corr(), annot=True, cmap='YlGnBu', fmt='.2f')
plt.title("Feature correlation heatmap")
plt.tight_layout()
plt.show()
```
