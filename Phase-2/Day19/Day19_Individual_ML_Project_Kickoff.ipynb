## 2️⃣ Dataset Understanding

| Field | Detail |
|---|---|
| **Dataset Name** | Crop Recommendation Dataset |
| **Dataset Source** | [kaggle.com/datasets/atharvaingle/crop-recommendation-dataset](https://www.kaggle.com/datasets/atharvaingle/crop-recommendation-dataset) |
| **Number of Rows** | 2,200 |
| **Number of Columns** | 8 (7 features + 1 target) |
| **Feature Columns** | N, P, K, temperature, humidity, ph, rainfall |
| **Target Variable** | `label` (22 crop classes) |

**Description of each feature**

| Feature | Description |
|---|---|
| N | Nitrogen content ratio in the soil |
| P | Phosphorous content ratio in the soil |
| K | Potassium content ratio in the soil |
| temperature | Ambient temperature in °C |
| humidity | Relative humidity (%) |
| ph | Soil pH value (acidity/alkalinity) |
| rainfall | Rainfall in mm |
| label | Target — the recommended crop out of 22 possible classes |
